# Max compat Ludo

```python
def compatLudo(w1,w2):
    return not (w1[0] < w2[0] <= w1[1]<w2[1] or w2[0] < w1[0] <= w2[1]<w1[1] or w1[1] < w2[1] <= w1[2] < w2[2] or w2[1] < w1[1] <= w2[2] < w1[2])
```

```python
def allWords(n):
    return [(i,j,k) for i in range(n) for j in range(i,n) for k in range(j,n)]
```

```python
words = allWords(3);words
```

```python
compat_edges = []
for i in range(len(words)):
    for j in range(i+1,len(words)):
        if compatLudo(words[i],words[j]):
            compat_edges.append([words[i],words[j]])
compat_graph = Graph([words,compat_edges]);compat_graph.show()
```
